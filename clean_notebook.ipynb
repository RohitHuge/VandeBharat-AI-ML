```python
import os
import yaml

# ===== FINAL CLASSES =====
final_classes = [
"air_brake_hose","air_spring","axle","wheel","battery_box","bio_tank",
"brake_system","cooling_grill","damper","disc","electrical_box",
"hydraulic_hose","ladder","suspension","pressure_valve","protective_grill",
"push_rod","sab","frame","transformer","pipe","reservoir","spring","bolt",
"crack","rust","broken","puncture","deformation","leakage","missing_part"
]

class_to_id = {c:i for i,c in enumerate(final_classes)}

# ===== MAPPING (IMPORTANT) =====
mapping = {

# battery
"Battery box": "battery_box",
"BatteryBox": "battery_box",
"Battery-box": "battery_box",
"battery_box": "battery_box",

# axle
"Axle": "axle",
"Axel Box": "axle",
"Axle Box Cover": "axle",

# wheel
"Wheel": "wheel",
"wheel": "wheel",

# electrical
"Electric Box": "electrical_box",
"Electrical box": "electrical_box",
"Electricity Box": "electrical_box",
"ElectricBox": "electrical_box",

# suspension
"Suspension": "suspension",
"Primary Suspension": "suspension",
"Secondary Suspension": "suspension",
"Secondary suspension": "suspension",

# bolts
"NutBoults": "bolt",
"Loose Nutboult": "bolt",
"Missing NutBoult": "bolt",
"missing_bolt": "bolt",

# defects
"Crack": "crack",
"crack": "crack",
"Rusting": "rust",
"rust": "rust",
"Broken": "broken",
"Puncture": "puncture",
"Deformation": "deformation",

# pipes
"pipe": "pipe",
"Pipes": "pipe",
"Outlet Pipe": "pipe",

# reservoir
"Reservoir": "reservoir",
"Water Reservoir": "reservoir",
"Water Reservior": "reservoir"
}

# ===== YOUR YAML PATHS =====
paths_yaml = [
"/content/drive/MyDrive/dataset/GX010982_frames_Soham/Labelled_images/component labeling new.yolov8/data.yaml",
"/content/drive/MyDrive/dataset/Labelled Frames 2.yolov8 (1)/data.yaml",
"/content/drive/MyDrive/dataset/Labelled Frames.yolov8 (1)/data.yaml",
"/content/drive/MyDrive/dataset/TrainDefectivePart.yolov8/data.yaml",
"/content/drive/MyDrive/dataset/TrainDefectivePart.yolov8/TrainDefectivePartsGuassianNoise.yolov8 (1)/data.yaml",
"/content/drive/MyDrive/dataset/TrainDefectiveParts/defect_dataset/data.yaml",
"/content/drive/MyDrive/dataset/TrainDefectsLabelling.yolov8 (1)/data.yaml",
"/content/drive/MyDrive/dataset/Yadnyesh_fixed/data.yaml",
"/content/drive/MyDrive/dataset/labelled frames/Labelled frames/data.yaml",
"/content/drive/MyDrive/dataset/train_fixed/data.yaml",
"/content/drive/MyDrive/dataset/train2_fixed/data.yaml"
]

# ===== FUNCTION TO FIX LABEL FILE =====
def process_label_file(label_path, old_names):
    new_lines = []

    with open(label_path, "r") as f:
        lines = f.readlines()

    for line in lines:
        parts = line.strip().split()
        if len(parts) < 5:
            continue

        old_id = int(parts[0])
        bbox = parts[1:]

        old_class = old_names[old_id]

        # Map to standardized name
        new_class = mapping.get(old_class, old_class.lower().replace(" ", "_"))

        if new_class not in class_to_id:
            continue  # skip unknown classes

        new_id = class_to_id[new_class]

        new_lines.append(f"{new_id} " + " ".join(bbox))

    with open(label_path, "w") as f:
        f.write("\n".join(new_lines))


# ===== MAIN LOOP =====
for yaml_path in paths_yaml:
    print(f"\nProcessing: {yaml_path}")

    if not os.path.exists(yaml_path):
        print("❌ YAML not found, skipping")
        continue

    try:
        with open(yaml_path) as f:
            data = yaml.safe_load(f)

        old_names = data["names"]

        # label folder
        label_dir = os.path.join(os.path.dirname(yaml_path), "train", "labels")

        if not os.path.exists(label_dir):
            print("❌ Labels folder not found, skipping")
            continue

        for file in os.listdir(label_dir):
            if file.endswith(".txt"):
                process_label_file(os.path.join(label_dir, file), old_names)

        print("✅ Done")

    except Exception as e:
        print(f"❌ Error: {e}")

print("\n🎉 ALL LABELS STANDARDIZED!")
```

```python
import os, shutil

# Paths
base = "/content/drive/MyDrive/final_dataset"
train3_img = "/content/drive/MyDrive/dataset/train 3/images"
train3_lbl = "/content/drive/MyDrive/dataset/train 3/labels"

pool_img = "/content/pool/images"
pool_lbl = "/content/pool/labels"

os.makedirs(pool_img, exist_ok=True)
os.makedirs(pool_lbl, exist_ok=True)

# Collect existing dataset
for split in ["train", "val"]:
    for f in os.listdir(f"{base}/{split}/images"):
        shutil.copy(f"{base}/{split}/images/{f}", pool_img)

    for f in os.listdir(f"{base}/{split}/labels"):
        shutil.copy(f"{base}/{split}/labels/{f}", pool_lbl)

# Add train3
for f in os.listdir(train3_img):
    shutil.copy(os.path.join(train3_img, f), pool_img)

for f in os.listdir(train3_lbl):
    shutil.copy(os.path.join(train3_lbl, f), pool_lbl)

print("✅ train3 merged into pool")
```

```python
import random

images = os.listdir(pool_img)
random.shuffle(images)

split = int(0.8 * len(images))

train_imgs = images[:split]
val_imgs = images[split:]

# Remove old split
shutil.rmtree(f"{base}/train")
shutil.rmtree(f"{base}/val")

# Recreate
for folder in ["train/images", "train/labels", "val/images", "val/labels"]:
    os.makedirs(f"{base}/{folder}", exist_ok=True)

def move(img_list, split_type):
    for img in img_list:
        lbl = img.rsplit(".", 1)[0] + ".txt"

        shutil.copy(os.path.join(pool_img, img),
                    f"{base}/{split_type}/images/{img}")

        shutil.copy(os.path.join(pool_lbl, lbl),
                    f"{base}/{split_type}/labels/{lbl}")

move(train_imgs, "train")
move(val_imgs, "val")

print("✅ train3 fully merged and dataset rebalanced")
```

```python
import cv2
from ultralytics import YOLO

# Load model
model = YOLO("/content/best.pt")   # adjust path if needed

# Open video
cap = cv2.VideoCapture("/content/input_video.mp4")

# Output video
out = cv2.VideoWriter(
    "/content/output_video.mp4",
    cv2.VideoWriter_fourcc(*'mp4v'),
    5,
    (int(cap.get(3)), int(cap.get(4)))
)

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # Run detection
    results = model(frame)

    # Draw boxes + labels
    annotated_frame = results[0].plot()

    # Save & show
    out.write(annotated_frame)
    cv2.imshow("Detection", annotated_frame)

    if cv2.waitKey(1) & 0xFF == 27:
        break

cap.release()
out.release()
cv2.destroyAllWindows()

print("✅ Detection video saved as output_video.mp4")
```

```python
import os
import cv2
from ultralytics import YOLO

# =========================
# PATHS (EDIT IF NEEDED)
# =========================
IMG_FOLDER = "/content/drive/MyDrive/final_dataset/val/images"
MODEL_PATH = "/content/best.pt"
OUTPUT_VIDEO = "/content/output_video.mp4"

# =========================
# LOAD MODEL
# =========================
model = YOLO(MODEL_PATH)

# =========================
# GET IMAGE LIST
# =========================
images = sorted(os.listdir(IMG_FOLDER))
images = [img for img in images if img.endswith(('.jpg', '.png', '.jpeg'))]

# =========================
# READ FIRST IMAGE (SIZE)
# =========================
first_img = cv2.imread(os.path.join(IMG_FOLDER, images[0]))
height, width, _ = first_img.shape

# =========================
# SET VIDEO WRITER
# =========================
FPS = 20  # you can change (10–30)

out = cv2.VideoWriter(
    OUTPUT_VIDEO,
    cv2.VideoWriter_fourcc(*'mp4v'),
    FPS,
    (width, height)
)

# =========================
# PROCESS IMAGES
# =========================
for img_name in images:
    img_path = os.path.join(IMG_FOLDER, img_name)

    frame = cv2.imread(img_path)

    # Run detection
    results = model(frame, conf=0.4)

    # Draw boxes + labels
    annotated_frame = results[0].plot()

    # Write frame to video
    out.write(annotated_frame)

# =========================
# RELEASE
# =========================
out.release()

print("✅ Done! Video saved at:", OUTPUT_VIDEO)
```
